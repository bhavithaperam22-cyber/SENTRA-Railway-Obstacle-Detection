```python
from ultralytics import YOLO
from google.colab.patches import cv2_imshow
import cv2

# Load your trained model
model = YOLO("/content/runs/detect/train-2/weights/best.pt")

# Path to the image you want to test
image_path = "/content/dataset/test/images/Animal-Obstacle-110-_jpg.rf.ad3cd0a4bb68ccbbef123f9b6d583780.jpg"

# Run prediction
results = model.predict(
    source=image_path,
    conf=0.25,        # Confidence threshold
    save=True,        # Save the result
    show_labels=True, # Show class names
    show_conf=True    # Show confidence scores
)

# Display the result
for r in results:
    img = r.plot()  # Get image with bounding boxes
    cv2_imshow(img) # Show the image
```
