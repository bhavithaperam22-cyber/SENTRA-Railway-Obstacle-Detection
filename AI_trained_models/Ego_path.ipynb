```python
import os
import shutil
import glob

print("="*60)
print("📂 ORGANIZING FILES")
print("="*60)

# Find the extracted ego path folder
repo_folder = None
for item in os.listdir('/content'):
    if os.path.isdir(os.path.join('/content', item)):
        if 'ego' in item.lower() or 'train' in item.lower():
            if 'zip' not in item:
                repo_folder = os.path.join('/content', item)
                print(f"✅ Found repository folder: {repo_folder}")
                break

# Find the extracted weights folder
weights_source = None
for root, dirs, files in os.walk('/content'):
    for dir_name in dirs:
        if dir_name == 'weights' and 'ego' not in root:
            weights_source = os.path.join(root, dir_name)
            print(f"✅ Found weights folder: {weights_source}")
            break
    if weights_source:
        break

# If we found both, copy weights into repository
if repo_folder and weights_source:
    # Create weights folder in repository
    dest_weights = os.path.join(repo_folder, 'weights')
    os.makedirs(dest_weights, exist_ok=True)

    # Copy all .pt files
    for file in os.listdir(weights_source):
        src = os.path.join(weights_source, file)
        dst = os.path.join(dest_weights, file)
        if os.path.isfile(src):
            shutil.copy(src, dst)
            print(f"   Copied: {file}")

    print(f"✅ Weights copied to: {dest_weights}")

    # Set the repo path
    repo_path = repo_folder
else:
    # If we couldn't find weights, check if they are already in the repo
    repo_path = "/content/train-ego-path-detection"
    if os.path.exists(repo_path):
        weights_path = os.path.join(repo_path, 'weights')
        if os.path.exists(weights_path):
            print(f"✅ Weights already in: {weights_path}")
            repo_path = repo_path
        else:
            print("❌ Could not find weights folder")
            print("📤 Please upload your best.pt file manually:")
            from google.colab import files
            uploaded = files.upload()
            os.makedirs(weights_path, exist_ok=True)
            for fname in uploaded.keys():
                shutil.move(f'/content/{fname}', f'{weights_path}/best.pt')
                print(f"✅ Uploaded: {fname}")
```

```python
import os
import shutil

print("="*60)
print("📂 FIXING NESTED FOLDER STRUCTURE")
print("="*60)

# The nested folder path
nested_path = "/content/train-ego-path-detection/train-ego-path-detection"

if os.path.exists(nested_path):
    print(f"✅ Found nested folder: {nested_path}")

    # Move all contents to the parent folder
    for item in os.listdir(nested_path):
        src = os.path.join(nested_path, item)
        dst = os.path.join("/content/train-ego-path-detection", item)

        # Remove destination if it exists
        if os.path.exists(dst):
            if os.path.isdir(dst):
                shutil.rmtree(dst)
            else:
                os.remove(dst)

        shutil.move(src, dst)
        print(f"   Moved: {item}")

    # Remove the now-empty nested folder
    shutil.rmtree(nested_path)
    print("✅ Removed nested folder")
else:
    print("❌ No nested folder found")
```

```python


%cd /content/train-ego-path-detection

import os
import sys
import cv2
import glob
import shutil
from google.colab.patches import cv2_imshow
from google.colab import files

print("="*60)
print("🔍 SENTRA - EGO-PATH DETECTION")
print("="*60)

# STEP 1: Upload image
print("\n📤 Upload railway track image:")
uploaded = files.upload()

# STEP 2: Find the uploaded file
original_name = list(uploaded.keys())[0]
print(f"📷 Uploaded: {original_name}")

# Search for the file (it might be in /content/ or /content/train-ego-path-detection/)
import glob
found_files = glob.glob(f'/content/**/{original_name}', recursive=True)
if not found_files:
    found_files = glob.glob(f'/content/**/*{original_name}*', recursive=True)

if not found_files:
    print("❌ File not found!")
    exit()

img_path = found_files[0]
img_name = os.path.basename(img_path)
print(f"📷 Using: {img_path}")

# STEP 3: Model
model_folder = "/content/train-ego-path-detection/weights/chromatic-laughter-5"

if not os.path.exists(model_folder):
    print(f"❌ Model not found: {model_folder}")
    print("\n📂 Available models:")
    for item in os.listdir("/content/train-ego-path-detection/weights"):
        if os.path.isdir(os.path.join("/content/train-ego-path-detection/weights", item)):
            print(f"   - {item}")
    exit()

print(f"✅ Using model: chromatic-laughter-5")

# STEP 4: Run detection
!mkdir -p /content/output
print(f"\n🔍 Processing: {img_name}")

# Use subprocess to handle paths properly
import subprocess
cmd = [
    "python", "detect.py",
    model_folder,
    img_path,
    "--output", "/content/output"
]

result = subprocess.run(cmd, capture_output=True, text=True, cwd="/content/train-ego-path-detection")

if result.stderr:
    print("⚠️ Errors:", result.stderr)

# STEP 5: Show results
print("\n📊 RESULTS:")
output_files = glob.glob('/content/output/*.jpg')
if output_files:
    for out_file in output_files:
        img = cv2.imread(out_file)
        if img is not None:
            print(f"✅ Output: {out_file}")
            cv2_imshow(img)
else:
    print("❌ No output generated")
    print("\n📂 Output folder contents:")
    !ls -la /content/output
```
